# Can we predict how ice deformation changes with the size of a square?
Run this notebook from its directory. It uses the 13 real derived observations, not the full satellite files. First explain why a sheet of ice moving east without changing shape can have speed but zero deformation.

```python
import json, math
from pathlib import Path
from pilot import evaluate
rows = json.loads(Path('results/observations.json').read_text())
first = rows[0]
print(first['start'], first['end'], first['scenarios']['primary']['target_per_day'])
```

## Write it before running it
If D = a × Lᵇ, then D₂/D₁ = (L₂/L₁)ᵇ. Taking logarithms gives b = ln(D₂/D₁) / ln(L₂/L₁). For the first real scene, write 0.05546 / 0.06173 = (5.6 / 2.8)ᵇ. Is b positive or negative? Why?

```python
d1, d2, d3 = first['scenarios']['primary']['target_per_day']
b = math.log(d2/d1)/math.log(2)
print('First two scales only: signed slope =', round(b, 3), 'beta =', round(-b, 3))
print('Prediction at 11.2 km:', round(d2 * 2**b, 5), 'measured:', round(d3, 5))
```

## Compare a fair baseline
M0: one level, one slope. M1: separate levels, one slope. M2: separate levels and slopes. Before running: why is M1 needed? What data must stay out of training? What would change your mind?

```python
for scenario in first['scenarios']:
    r = evaluate(rows, scenario)
    print(scenario, {k:round(v, 4) for k,v in r['mean_held_out_log_mse'].items()})
    print('Weeks in which separate slopes beat separate levels:', r['weeks_slopes_beat_levels'], '/', r['n_scored'])
```
